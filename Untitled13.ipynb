```python
import numpy as np, rasterio, matplotlib
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from landcover import config
from landcover.preprocess import load_s2_features
from landcover.labels import remap_worldcover

feats, valid, _ = load_s2_features("data/raw/nagpur_s2.tif")
with rasterio.open("data/raw/nagpur_label.tif") as s:
    ref = remap_worldcover(s.read(1))
with rasterio.open("outputs/nagpur_classified.tif") as s:
    pred = s.read(1)

h = min(ref.shape[0], pred.shape[0], feats.shape[0])
w = min(ref.shape[1], pred.shape[1], feats.shape[1])
ref, pred, feats = ref[:h, :w], pred[:h, :w], feats[:h, :w]

both = (ref != 255) & (pred != 255)
print(f"pixel agreement with reference labels: {(ref == pred)[both].mean():.3f}")

rgb = feats[..., [2, 1, 0]]
lo, hi = np.percentile(rgb[rgb > 0], (2, 98))
rgb = np.clip((rgb - lo) / (hi - lo), 0, 1)

cmap = ListedColormap(config.CLASS_COLORS)
step = 3
fig, ax = plt.subplots(1, 3, figsize=(18, 6))
ax[0].imshow(rgb[::step, ::step]); ax[0].set_title("Sentinel-2 true colour")
ax[1].imshow(np.ma.masked_equal(ref, 255)[::step, ::step], cmap=cmap, vmin=0, vmax=4, interpolation="nearest")
ax[1].set_title("Reference labels (WorldCover)")
ax[2].imshow(np.ma.masked_equal(pred, 255)[::step, ::step], cmap=cmap, vmin=0, vmax=4, interpolation="nearest")
ax[2].set_title("U-Net prediction")
for a in ax: a.axis("off")
handles = [matplotlib.patches.Patch(color=c, label=n) for n, c in zip(config.CLASS_NAMES, config.CLASS_COLORS)]
fig.legend(handles=handles, loc="lower center", ncol=5)
plt.savefig("outputs/nagpur_comparison.png", dpi=150, bbox_inches="tight")
plt.show()
```
